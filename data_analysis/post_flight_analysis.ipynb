# CELLHAWK Post-Flight Analysis
This notebook ingests EKF telemetry logs and computes field campaign success metrics against simulation baselines.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import ttest_1samp, ks_2samp

# Load telemetry
# df = pd.read_csv('logs/flight_01.csv')
# For demonstration, we use a mock DataFrame representing the parsed logs
df = pd.DataFrame({
    'timestamp': np.linspace(0, 60, 600),
    'ekf_x': np.random.normal(0, 1.2, 600),
    'ekf_y': np.random.normal(0, 1.2, 600),
    'gps_x': np.zeros(600),
    'gps_y': np.zeros(600),
    'tier': np.ones(600), # Tier 1
    'handover_latency_ms': np.zeros(600),
    'bandwidth_kbps': np.random.uniform(1.5, 3.5, 600)
})
df.head()
```

## 1. RMS Position Error vs Ground Truth

```python
def calculate_rms(df):
    error_x = df['ekf_x'] - df['gps_x']
    error_y = df['ekf_y'] - df['gps_y']
    rms = np.sqrt(np.mean(error_x**2 + error_y**2))
    return rms

rms_error = calculate_rms(df)
print(f"Baseline GPS RMS Error: {rms_error:.2f} meters")
```

## 2. Bandwidth Utilization

```python
mean_bw = df['bandwidth_kbps'].mean()
peak_bw = df['bandwidth_kbps'].max()
print(f"Mean Bandwidth: {mean_bw:.2f} kbps/drone")
print(f"Peak Bandwidth: {peak_bw:.2f} kbps/drone")
```
